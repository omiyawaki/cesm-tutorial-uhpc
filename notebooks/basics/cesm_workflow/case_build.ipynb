cd $MYROOT/cases/f2000.test
```
    
<br>
    
<strong>Example `case.build` command: </strong><br>    
    
```
./case.build
```
    
</div>



<div class="alert alert-info">  
<details>

<summary> <font face="Times New Roman" color='blue'>Click here for example output</font> </summary>

<br>
    
```
