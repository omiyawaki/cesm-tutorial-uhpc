# Model Output



```{admonition} New to the command line?
:class: tip
- `ls -l <folder>` lists files with their sizes. `du -sh <folder>` prints the total size of a folder, for example `du -sh /usr/local/research/miyawaki/users/$USER/archive`.
- You can also browse these folders in the OnDemand **Files** app by typing the path into its path box.
- `ncdump -h <file>.nc` prints the list of variables in a NetCDF file (run `source /usr/local/research/miyawaki/opt/cesm-deps/ENV.sh` first).
```


If the model run is successful, the CESM netcdf output history files are automatically moved to a short term archive (`$DOUT_S_ROOT`) by `case.st_archive`.

![Case Submit Image](../../../images/basics/CESM2_short_term_archive.png)

**Notes:**
- If a model run was unsuccessful the output remains in the Run Directory (`$RUNDIR`) and the short term archive is not created.
- On NCAR HPC both `$RUNDIR` and `$DOUT_S_ROOT` are in scratch space that is scrubbed after a number of days. On UHPC they are planned to be in your folder under `/usr/local/research/miyawaki/users/$USER` on the shared research drive, which is not scrubbed. The whole group shares 4.9 TB, so delete run directories you no longer need. A soft limit of about 300 GB per student is planned (**TODO:** not yet decided). Never leave output on a compute node's local `/scratch` disk.
