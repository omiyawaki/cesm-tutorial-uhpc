# Explore CESM Code


The shared CESM code is already on UHPC. On this page you look around its folders. The commands here only list files and change nothing, so you can type them in the OnDemand Shell tab. Specifically in this exercise we will:

- Step 1. Locate the CESM Code area on the UHPC file system.
- Step 2. Examine the CIME Directory area.
- Step 3. Examine the Model Components area.
- Step 4. Look at the Community Atmosphere Model (CAM) Component Model.

# Step 2. Examine the CIME Directory

In the OnDemand Shell tab, move to the CIME directory and examine the contents. Remember that CIME is the Common Infrastructure for Modeling the Earth. It is a python-based framework that integrates the various components of the CESM into a single modeling framework. The CIME Scripts directory will be the location where you will start the Workflow of creating a new case through the rest of the Tutorial.

<div class="alert alert-info">
