# Exercise Overview 


## Learning Goals:

- The student will successfully set up and run their first (short) CESM simulation.
- This exercise uses the CESM workflow information presented in this chapter.

## Exercise overview:

By the end of this exercise you will have:
- created, set up, built, run, and examined the output of a short atmosphere-only `F2000climo` case.

```{note}
**Union HPC:** NCAR's original exercises here used the fully coupled `B1850` case. In this edition, your first run is the
[UHPC quickstart](../uhpc/quickstart.md): follow it from start to finish, then come back to the pages below.
You do not need to download CESM with git; everyone uses the shared CESM2.1.5 copy on UHPC.
```
