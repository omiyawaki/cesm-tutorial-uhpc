# Basics

```{note}
**Read this chapter after the quickstart.** You already made, built and ran your first case, `f2000.test`, in the [UHPC quickstart](../uhpc/quickstart.md). This chapter explains those steps. On UHPC the one-time setup (downloading CESM, input data and porting) is already done for you.
```


The CESM Model is developed around a structured workflow. The first step of this workflow is to set up your workspace, which is usually a one-time task. After setting up your workspace, a basic CESM simulation can be run with only four commands. This section will cover these steps and enable you to complete your first CESM run.

## Workflow Elements

<div class="alert alert-info" style="text-align: left;">

- Set up workspace (one time setup)
    - **``Download``** the CESM code
    - **``Create or Locate ``** an **``Input Data``** Root Directory
    - Possibly **``Porting``** if not on a setup machine
- Creating, Running and Reviewing a Case
    - **``Create``** a new case
    - **``Invoke``** `case.setup`
    - **``Build``** the executable with `case.build`
    - **``Run``** the model with `case.submit`
    - **``Review``** output data

</div>

## Setting up your environment

Every time you log in to UHPC, the CESM environment must be loaded before you build a case. The `source .../ENV.sh` line you added to `~/.bashrc` in step 3 of the [UHPC quickstart](../uhpc/quickstart.md) does this at every login. To check, type `echo $CESMROOT`. It should print `/usr/local/research/miyawaki/cesm/cesm2`.
