# Exercise Overview

This exercise is to introduce you to the UHPC computing system and set up your environment. Specifically, in this exercise we will:

- Step 1. Log in to UHPC.
- Step 2. Examine the UHPC file system.



## Step 1: UHPC Login

On UHPC you use Open OnDemand in your web browser instead of logging in from a terminal. <br>

<div class="alert alert-info">
<strong>Login to UHPC: </strong><br><br>

```
https://ondemand.hpc.union.edu
```

</div>

Connect to the Union VPN first. Then sign in to <https://ondemand.hpc.union.edu> with your cluster username and password and open **Clusters** → **HPC Cluster Shell Access** (see [Open OnDemand and shell basics](../uhpc/ondemand_basics.md)). For login problems, see [When something goes wrong](../uhpc/when_things_go_wrong.md).

On NCAR's own systems, users contact user support at the NCAR Computational and Information Systems Laboratory (CISL), linked below. That link does not cover UHPC.

<div class="alert alert-info" style="text-align: center;">

[CISL User Support](https://ithelp.ucar.edu/plugins/servlet/desk/site/rc)

</div>

## Step 2: Look Around the UHPC File System

Now that you are on UHPC you can start to explore the file system. The following simple exercises will start that process.

<div class="alert alert-info">
<strong>Show your current directory: </strong><br>

```
pwd
```
<br>

<strong>List the shared research directory, which holds the CESM code and input data: </strong><br>

```
ls /usr/local/research/miyawaki
```
<br>
    
<strong>List the contents of the CESM code directory: </strong><br>

```
ls /usr/local/research/miyawaki/cesm/cesm2
```

</div>

The input data directory (`DIN_LOC_ROOT`) is `/usr/local/research/miyawaki/cesm/inputdata`. Your own folder is `/usr/local/research/miyawaki/users/$USER`.



Feel free to look around more of the UHPC system. See the [UHPC quickstart](../uhpc/quickstart.md) and [Differences from the NCAR tutorial](../uhpc/differences.md) for where things live.
