## Union HPC (UHPC)

### Logging in to UHPC

On UHPC you log in through Open OnDemand in your web browser, not with `ssh`. You do not need X11 forwarding, which NCAR users set up for graphical programs such as `ncview`. Those programs are not installed on UHPC, and you look at model output in the OnDemand Jupyter app instead. Logging in is explained in the [UHPC quickstart](../uhpc/quickstart.md).

```
https://ondemand.hpc.union.edu   (Shell app)
```

Use campus Wi-Fi or the Union VPN. See [Open OnDemand](../uhpc/ondemand_basics.md).

## Modules and software on UHPC

NCAR provides programs on its HPC systems that you add to your environment with the ``module load`` command. **The NCAR modules do not exist on UHPC.** On UHPC, CESM's compilers, MPI and NetCDF libraries are set up by an `ENV.sh` script (see the [UHPC quickstart](../uhpc/quickstart.md)). UHPC does use `module` for a few packages (`module avail` lists them, for example `python/miniforge`). NCO, CDO, NCL and `ncview` are not installed. Use Python instead.

On UHPC, the `ENV.sh` line from the quickstart loads every module CESM needs, so this tutorial does not ask you to load any other module. The commands below show what is loaded and what exists.

To learn more about modules on NCAR HPC, follow the link below.

<div class="alert alert-info" style="text-align: center;">

[CISL documentation on modules](https://arc.ucar.edu/knowledge_base/72581272)

</div>

### Checking what modules are loaded in your environment

To check what modules you have loaded in your environment, use the following at the command line:

````bash
module list
````

### Checking what modules are available

You can see which modules are available on the HPC resources by using the following at the command line:

````bash
module avail
````

The command prints a list of the modules available to load to your environment.

![Modules](../../images/prereqs/modules.png)

*<p style="text-align: center;"> Figure: Some modules available on NCAR HPC resources (UHPC will look different). D: Default Module. L: Module is loaded. </p>*

The modules highlighted in blue are text editors discussed further in the [text editors section](../resources/text_editors.ipynb).

The modules highlighted in red are netCDF tools that will be discussed further in the [netCDF section](../resources/netcdf.ipynb).

Note that on NCAR HPC there are many other analysis software tools available to load into your environment. These include: idl, julia, matlab, R, python, ncl, etc. While some of these tools are discussed in the [analysis tools section](../diagnostics/additional/analysis_tools.ipynb), we do not highlight them all here and encourage you to further explore your environment and needs.

### Loading a new module to your environment

On NCAR HPC you load a module with `module load <name>`, for example `module load python/miniforge`. `ncview` does not exist on UHPC, so do not type that line. On UHPC the one extra module you might use later is `python/miniforge`:

````bash
module load python/miniforge
````
