```
    cd /usr/local/research/miyawaki/cesm/cesm2/cime/scripts
    ./create_newcase --case /usr/local/research/miyawaki/users/$USER/cases/f.day2.1 --res f19_f19_mg17 --compset F2000climo --output-root /usr/local/research/miyawaki/users/$USER/scratch
```
<br>
    
In your case directory, change debugging levels and set runtime variables:

```
    cd /usr/local/research/miyawaki/users/$USER/cases/f.day2.1
    ./xmlchange INFO_DBUG=2,STOP_N=1,STOP_OPTION=nmonths
    ./xmlchange --subgroup case.run JOB_WALLCLOCK_TIME=4:00:00
```
<br>    
    
Remember to manually update your README.case file to document your changes.


Carry on to setup, build and submit the run:
    
```
      ./xmlchange DOUT_S_ROOT=/usr/local/research/miyawaki/users/$USER/archive/f.day2.1
      ./case.setup 
      ./case.build
      ./case.submit -a "--exclude=hpcc03,hpcc04"
```

On UHPC, `./case.build` is run without `qcmd`, in the OnDemand Shell inside tmux.


</details>
</div>

<div class="alert alert-success">
   
<details>   
   <summary>
        <img src="https://freesvg.org/img/help-browser.png" width="20px">
       <font face="Times New Roman" color='blue' size="4">Click here for the solution to Part II</font>
    </summary>

<br>
1. To continue the run, set `CONTINUE_RUN` to `TRUE`
    
```
    ./xmlchange CONTINUE_RUN=TRUE
```

<br>
2. Keep `STOP_OPTION` as "nmonths” 
    
    
<br>    
3. Set `STOP_N` to “2”
    
```
    ./xmlchange STOP_N=2
```

<br>
4. Set INFO_DBUG to 1
    
```
    ./xmlchange INFO_DBUG=1
```

<br>
5. Change the wallclock time to 12 hours, enough for 2 more model months (the UHPC maximum is 3 days):
    
```
    ./xmlchange --subgroup case.run JOB_WALLCLOCK_TIME=12:00:00
    ./xmlchange --subgroup case.st_archive JOB_WALLCLOCK_TIME=6:00:00
```

<br>
6. Submit (./case.submit) from your f.day2.1 case directory:

```
    ./case.submit -a "--exclude=hpcc03,hpcc04"
```
    
</details>
</div>
