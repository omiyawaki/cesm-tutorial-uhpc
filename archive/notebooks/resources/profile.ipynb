# UHPC Environment

The NCAR version of this page described the 2023 in-person tutorial: a tutorial project account, daily reservation queues on Cheyenne, and `.profile`/`.tcshrc` files that load NCAR modules. **None of these exist on UHPC.** This page describes the UHPC equivalents instead.

## Accounts, projects and queues

- Your UHPC account is created by the Union HPC administrators (hpcadmin@union.edu). You also need to be in the `res-miyawaki` group to read the shared CESM files. **TODO:** confirm how to check this.

- There are **no reservation queues**. Use the Slurm partition `hpcc` for CESM runs:

## Setting up your UHPC environment

Every time you log in to UHPC, make sure the CESM environment is loaded before you create, build or submit a case.

### SHELL environment

To determine what shell you are using, type:

```
echo $SHELL
```

If it returns `/bin/bash`, you are a bash user. The commands in this edition assume bash. **TODO:** confirm the default shell for UHPC accounts.
